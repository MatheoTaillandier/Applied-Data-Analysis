```python
import os
import pandas as pd
```

```python
DATA_PATH = "./data/"
interactive_data = pd.read_csv(os.path.join(DATA_PATH, "interactive_data.csv"))
print(len(interactive_data))
interactive_data.head()
```

```python
#drop weird additional indexing column 
interactive_data_clean = interactive_data.drop(columns=['Unnamed: 0'], axis=1)
print(interactive_data_clean.columns)
print(interactive_data_clean[interactive_data_clean.Intent=="Unknown"].count())
interactive_data_clean = interactive_data_clean[(interactive_data_clean.Intent!="Unknown") & (interactive_data_clean.Intent!="None selected")].reset_index(drop=True)
print(len(interactive_data_clean))
```

```python
print(interactive_data_clean.Intent.unique())
```

```python
ratio_suicide = interactive_data_clean[interactive_data_clean.Intent=="Suicide"]['Deaths'].sum()/interactive_data_clean.Deaths.sum()
ratio_homicide = interactive_data_clean[interactive_data_clean.Intent=="Homicide"]['Deaths'].sum()/interactive_data_clean.Deaths.sum()
ratio_accident = interactive_data_clean[interactive_data_clean.Intent=="Accident"]['Deaths'].sum()/interactive_data_clean.Deaths.sum()

print(f"{ratio_suicide} of all gun deaths are suicides")
print(f"{ratio_homicide} of all gun deaths are homicides")
print(f"{ratio_accident} of all gun deaths are accidents")
```

```python
suicide_interactive_data = interactive_data_clean[interactive_data_clean.Intent == "Suicide"]
suicide_interactive_data = suicide_interactive_data[suicide_interactive_data.Gender != "None selected"]
ratio_suicide_male = suicide_interactive_data[suicide_interactive_data.Gender == "Male"].Deaths.sum()/suicide_interactive_data.Deaths.sum()

print(f"{ratio_suicide_male} of suicide vitims are male")
```

```python
homicide_interactive_data_race_age = interactive_data_clean[interactive_data_clean.Intent == "Homicide"]
homicide_interactive_data_race_age = homicide_interactive_data_race_age[(homicide_interactive_data_race_age.Gender != "None selected")\
                                                       & (homicide_interactive_data_race_age.Age == "15 - 34")\
                                                        & (homicide_interactive_data_race_age.Race != "None selected")]
ratio_black = homicide_interactive_data_race_age[homicide_interactive_data_race_age.Race == "Black"].Deaths.sum()/homicide_interactive_data_race_age.Deaths.sum()
print(f"{ratio_black} of homicide victims who are males in the age-group of 15--34 are black.")
```

```python
homicide_interactive_data = interactive_data_clean[interactive_data_clean.Intent == "Homicide"]
homicide_interactive_data = homicide_interactive_data[(homicide_interactive_data.Gender != "None selected")]

women_ratio = homicide_interactive_data[homicide_interactive_data.Gender == "Female"].Deaths.sum()/homicide_interactive_data.Deaths.sum()

print(f"Women constitue only {women_ratio} of the total homicide victims")
```
